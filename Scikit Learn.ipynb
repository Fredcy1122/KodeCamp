```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures
from sklearn.ensemble import (HistGradientBoostingRegressor, RandomForestRegressor)
```

```python
housing = datasets.fetch_california_housing()
```

```python
x = housing.data
y = housing.target
print(housing.feature_names)
print(x[0])
print(y[0])
```

```python
poly = PolynomialFeatures()
x = poly.fit_transform(x)
x.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state= 432)

print(y_train.size)
print(housing.feature_names)
print(x_train[0])
print(y_train[0])
```

lr = LinearRegression()
gr = HistGradientBoostingRegressor()
rf = RandomForestRegressor( n_jobs = -1)

for i in [lr, gr, rf]:
    i.fit(x_train, y_train)
    y_pred = i.predict(x_test)

    score = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    print (i, score)
    print(i, mse)


```python
for j in [0.1, 0.05, 0.001, 0.0001]:
    for i in [200, 400, 500]:
        model = HistGradientBoostingRegressor(max_iter = i)
        model.fit(x_train, y_train)
        y_pred = model.predict(x_test)

        score = r2_score(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)

        print ( i, j, score, mse)
```
