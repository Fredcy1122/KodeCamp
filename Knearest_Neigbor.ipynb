```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
# 1. Generate synthetic data with 4 clusters
# We are creating 300 data points grouped around 4 hidden centers
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.40, random_state=0)

# Plot the raw, unclustered, and unlabeled data
plt.scatter(X[:, 0], X[:, 1], s=50, color='gray')
plt.title("Mystery Data: Unlabeled Observations")
plt.xlabel("Feature A")
plt.ylabel("Feature B")
plt.show()
```

```python
from sklearn.cluster import KMeans

# 1. Initialize the model with 4 clusters
# We are telling the algorithm to look for 4 clusters
kmeans = KMeans(n_clusters=4, n_init='auto')

# 2. Train the model on our data
kmeans.fit(X)
y_kmeans = kmeans.predict(X)

# 3. Plot the clustered data and the Centroids
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='plasma')

centers = kmeans.cluster_centers_

# 4. Visualize the cluster assignments and centroids as big red stars
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=250, alpha=0.9, marker='*')
plt.title("K-Means Results: Clusters and Centroids")
plt.show()
```
